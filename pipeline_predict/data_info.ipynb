```python
import cloudpickle
def load_pipeline(location): 
    with open(location, 'rb') as filename: 
        pipeline = cloudpickle.load(filename)

    return pipeline 
```

```python
pipeline = load_pipeline('model_info/data_processing/application_record_pipeline.pkl')
pipeline
```

```python
transformed_df = pipeline.transform(df)
transformed_df
```

```python
import torch 
def load_model(location): 
    model_state  = torch.load(location, weights_only = False )
    print('model loaded in from: ', location)
    return model_state
```
